```python
# Group data by target pathway to understand drug response at pathway level
# Calculate:
# - mean LN_IC50 -> average drug sensitivity
# - median LN_IC50 -> measure of typical sensitivity 
# - std -> variability in drug response across cell lines

# Sort pathways by median LN_IC50
#  lower median = higher sensitivity
#  helps rank pathways from most to least sensitive
pathway_stats = gdsc.groupby("TARGET_PATHWAY")["IC50"].agg(
    mean="mean",
    median="median",
    std="std"
).sort_values(by="median")


low_median = pathway_stats["median"].quantile(0.25)
high_std = pathway_stats["std"].quantile(0.75)

important_pathways = pathway_stats[
    (pathway_stats["median"] <= low_median) &
    (pathway_stats["std"] >= high_std)
]




import matplotlib.pyplot as plt
import seaborn as sns


sns.set(style="white")

plt.figure(figsize=(8,6))

# Scatter plot
# Each point = one pathway
# x-axis = sensitivity (median LN_IC50)
# y-axis = variability (std)
sns.scatterplot(
    data=pathway_stats,
    x="median",
    y="std",
    s=90,
    color="black"
)

# Thresholds
# 25th percentile → sensitive pathways
# 75th percentile → highly variable pathways
median_thresh = pathway_stats["median"].quantile(0.25)
std_thresh = pathway_stats["std"].quantile(0.75)

# Divide plot into 4 quadrants based on thresholds
plt.axvline(median_thresh, linestyle="--")
plt.axhline(std_thresh, linestyle="--")


# Label the pathways
for pathway in pathway_stats.index:
    plt.text(
        pathway_stats.loc[pathway, "median"] + 0.05,
        pathway_stats.loc[pathway, "std"] + 0.05,
        pathway,
        fontsize=8,
        weight="bold"
    )

# Axis labels
plt.xlabel("Median LN_IC50 (Sensitivity)")
plt.ylabel("Standard Deviation (Variability)")

# Title
plt.title("Pathway-Level Drug Sensitivity and Variability")

# Remove top/right spines (clean look)
#sns.despine()

plt.tight_layout()
plt.show()
```

```python
def compute_effect(data, feature, pos_label='Y', neg_label='N'):
    effect = (
        data
        .groupby(["DRUG", feature])["IC50"]
        .mean()
        .unstack()
    )

    # Handle missing columns safely
    if pos_label in effect.columns and neg_label in effect.columns:
        effect["diff"] = effect[pos_label] - effect[neg_label]
    else:
        effect["diff"] = None

    return effect["diff"]

gdsc.columns

cna = compute_effect(subset, "CNA")
msi = compute_effect(subset, "MSI_STATUS", pos_label='MSI-H', neg_label='MSS/MSI-L')
expr = compute_effect(subset, "GENE_EXPR")
meth = compute_effect(subset, "METHYLATION")

combined = pd.DataFrame({
    "CNA": cna,
    "MSI": msi,
    "Expression": expr,
    "Methylation": meth
})
# Fill NaN values with 0 immediately after creation
combined = combined.fillna(0)
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# ---- Step 1: Ensure consistent ordering (same as heatmap) ----
combined_sorted = combined.loc[
    combined.mean(axis=1).sort_values().index
]

# ---- Step 2: Convert to long format ----
plot_df = combined_sorted.rename_axis("Drug").reset_index().melt(
    id_vars="Drug",
    var_name="Feature",
    value_name="Effect"
)

# ---- Step 3: Define symmetric color scale ----
vmax = np.abs(plot_df["Effect"]).max()
vmin = -vmax

# ---- Step 4: Plot ----
plt.figure(figsize=(10, 8))

scatter = plt.boxplot(
    x=plot_df["Feature"],
    y=plot_df["Drug"],
    c=plot_df["Effect"],
    cmap="RdBu_r",
    vmin=vmin,
    vmax=vmax,
    s=np.abs(plot_df["Effect"]) * 200 + 20,  # size reflects magnitude
    edgecolor="black",
    linewidth=0.5
)

# ---- Step 5: Colorbar ----
cbar = plt.colorbar(scatter)
cbar.set_label("Effect on IC50 (Δ)", fontsize=11)

# ---- Step 6: Labels & styling ----
plt.title(
    "Genomic Feature Effects on Drug Sensitivity\n(Effect Size & Direction)",
    fontsize=14,
    weight="bold"
)

plt.xlabel("Genomic Feature", fontsize=12)
plt.ylabel("Drug", fontsize=12)

plt.xticks(rotation=0)
plt.yticks(fontsize=9)

plt.grid(axis='x', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```
